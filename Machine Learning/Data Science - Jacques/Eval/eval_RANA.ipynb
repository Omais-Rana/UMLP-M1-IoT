```python
# Code Explanation
# dataset = loadtxt('example.csv', delimiter=','): This loads the numerical data from the 'example.csv' file using a comma as a separator.
# X = dataset[:,0:8]: Extracts all rows from the dataset and selects the first eight columns for input variables.
# y = dataset[:,8]: Extracts all rows from the dataset and selects the ninth column for output class variable. 
```

```python
from keras.models import Sequential
from keras.layers import Dense, Input
model = Sequential([
    Input(shape=(8,)), # model expects rows of data with 8 variables
    Dense(12, activation='relu'), # first hidden layer has 12 nodes and uses the relu activation function
    Dense(8, activation='relu'), # second hidden layer has 8 nodes and uses the relu activation function
    Dense(1, activation='sigmoid') # output layer has one node and uses the sigmoid activation function
])
```

```python
# Code Explanation
# loss='binary_crossentropy': It is the loss function used for binary classification problems where the output is a class of 0 or 1.
# optimizer='adam': It specifies the Adam optimization algorithm to automatically manage and update the model weights during training.
# metrics=['accuracy']: It instructs the model to report the percentage of accuracy as it trains. 
```

```python
model.fit(X, y, epochs=200, batch_size=10) # fit() function call
```

```python
# Definitions
# Epochs: The total number of times the learning algorithm will study the entire training dataset.  
# Batch Size: The number of samples processed by the network before the weights are updated.
```

```python
loss, accuracy = model.evaluate(X, y)
print(f"Accuracy: {accuracy*100:.2f}%") # evaluate and print the accuracy
```

```python
# Complete model training and evaluation
import numpy as np
from keras.models import Sequential
from keras.layers import Dense, Input

raw_data = """6,148,72,35,0,33.6,0.627,50,1
1,85,66,29,0,26.6,0.351,31,0
8,183,64,0,0,23.3,0.672,32,1
1,89,66,23,94,28.1,0.167,21,0
0,137,40,35,168,43.1,2.288,33,1"""

with open('example.csv', 'w') as f:
    f.write(raw_data)

dataset = np.loadtxt('example.csv', delimiter=',')
X = dataset[:, 0:8]
y = dataset[:, 8]

model = Sequential([
    Input(shape=(8,)), 
    Dense(12, activation='relu'),
    Dense(8, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy']) 

model.fit(X, y, epochs=200, batch_size=10) 

loss, accuracy = model.evaluate(X, y) # [cite: 910]
print(f"\nFinal Accuracy: {accuracy * 100:.2f}%")
```
