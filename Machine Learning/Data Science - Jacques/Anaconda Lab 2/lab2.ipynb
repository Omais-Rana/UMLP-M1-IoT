```python
import numpy as np
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression

x, y = make_regression(n_samples=100, n_features=3, noise=10)

print("First 5 values of x (Features):\n", x[:5])
print("\nFirst 5 values of y (Targets):\n", y[:5])

# Create an instance of the Linear Regression class
model_linReg = LinearRegression()

# Train the model on our data set using the fit() method
model_linReg.fit(x, y)

# Make predictions using the trained model
predictions = model_linReg.predict(x)

accuracy = model_linReg.score(x, y)
print(f"\nModel Accuracy Score: {accuracy:.4f}")
```
